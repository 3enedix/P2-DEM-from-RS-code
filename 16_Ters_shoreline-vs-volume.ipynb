```python
poly_all = pickle.load(open(path_rtsdtm + f'poly_all_{epsg}.pkl', 'rb'))
poly_west = pickle.load(open(path_rtsdtm + f'poly_west_{epsg}.pkl', 'rb'))
poly_center = pickle.load(open(path_rtsdtm + f'poly_center{epsg}.pkl', 'rb'))
poly_east = pickle.load(open(path_rtsdtm + f'poly_east{epsg}.pkl', 'rb'))

poly_all = CD_geometry.transform_polygon(poly_all, epsg, epsg_local)
poly_west = CD_geometry.transform_polygon(poly_west, epsg, epsg_local)
poly_center = CD_geometry.transform_polygon(poly_center, epsg, epsg_local)
poly_east = CD_geometry.transform_polygon(poly_east, epsg, epsg_local)
```

```python
# RTS-DTM volume changes
volumes = pickle.load(open(path_rtsdtm+'volumes.pkl', 'rb'))
```

## Correlation
V2: Compute statistics for each transect (no spatial averaging before)  
e.g. correlation between cross-shore time series of each transect against one volume change time series  

```python
cd_tcorr_df = pd.DataFrame.from_dict(cd_tcorr).set_index(c_dates)
cd_tcorr_yearly_av = cd_tcorr_df.groupby(pd.Grouper(freq='YE')).mean()
cd_tcorr_yearly_av = cd_tcorr_yearly_av.set_index(cd_tcorr_yearly_av.index.year)
cd_tcorr_yearly_av = cd_tcorr_yearly_av.drop([1992, 2001, 2002, 2003, 2012])
```

```python
# Correlation for each transect
corr = pd.DataFrame(columns=volumes.columns, index=cd_tcorr_yearly_av.columns)

for transect in cd_tcorr_yearly_av.columns:
    for col in corr.columns:
        not_nan = np.logical_or(cd_tcorr_yearly_av[transect].isna(), volumes[col].isna())
        corr.loc[transect, col], _ = stats.pearsonr(cd_tcorr_yearly_av.loc[~not_nan, transect], volumes.loc[~not_nan, col])
        # corr.loc[transect, col], _ = stats.spearmanr(cd_tcorr_yearly_av.loc[~not_nan, transect], volumes.loc[~not_nan, col])
```

```python
idx_all = transects.reset_index(drop=True).intersects(poly_all)
idx_west = transects.reset_index(drop=True).intersects(poly_west)
idx_center = transects.reset_index(drop=True).intersects(poly_center)
idx_east = transects.reset_index(drop=True).intersects(poly_east)
```

```python
c_vali = 'grey'
c_kf = 'brown'
c_rts = 'orange'
```

```python
def plot_corr_perTransect(ax, idx, area, title_area, legend=False, ylabel=False):
    ax.plot(corr.loc[idx, f'jarkus_{area}'], '.-', c=c_vali, label='JARKUS')
    ax.plot(corr.loc[idx, f'kf_{area}'], '.-', c=c_kf, label='KF-DTM')
    ax.plot(corr.loc[idx, f'rts_{area}'], '.-', c=c_rts, label='RTS-DTM')
    if legend:
        ax.legend(bbox_to_anchor=(1,1))
    ax.set_xlabel('Transect')
    if ylabel:
        ax.set_ylabel('Correlation')
    ax.set_ylim([-1,1])
    ax.set_title(f'{title_area} coastline')
    ax.grid()
```

```python
fig, axs = plt.subplots(1,4,figsize=(20,5))
fig.tight_layout()
fig.suptitle('Shoreline changes vs. volume changes', y=1.1)
plot_corr_perTransect(axs[0], idx_all, area='all', title_area='Entire', ylabel=True)
plot_corr_perTransect(axs[1], idx_west, area='west', title_area='Western')
plot_corr_perTransect(axs[2], idx_center, area='center', title_area='Centre')
plot_corr_perTransect(axs[3], idx_east, area='east', title_area='Eastern', legend=True)
```

## Trends

```python
trends_cassie = pd.DataFrame(columns=['trend'], index=cd_tcorr_yearly_av.columns)

for transect in cd_tcorr_yearly_av.columns:
    if cd_tcorr_yearly_av[transect].isna().sum() > len(cd_tcorr_yearly_av-5): # min 4 non-nan values
        continue
    trends_cassie.loc[transect, 'trend'], _, _ = CD_statistics.compute_trend_with_error(cd_tcorr_yearly_av.index.values, cd_tcorr_yearly_av[transect])
```

```python
trends_vol = pd.DataFrame(index=volumes.columns)
for col in volumes.columns:
    trends_vol.loc[col, 'trend'], _, _ = CD_statistics.compute_trend_with_error(volumes.index.values, volumes[col].values)
```

```python
print('Mean trends CASSIE cross-shore shoreline time series:')
mean_entire = trends_cassie.loc[idx_all,'trend'].mean()
print(f'Entire coastline: {round(mean_entire,2)} m/year')
mean_west = trends_cassie.loc[idx_west,'trend'].mean()
print(f'West coastline: {round(mean_west,2)} m/year')
mean_center = trends_cassie.loc[idx_center,'trend'].mean()
print(f'Center coastline: {round(mean_center,2)} m/year')
mean_east = trends_cassie.loc[idx_east,'trend'].mean()
print(f'East coastline: {round(mean_east,2)} m/year')
```
